```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
    .master('local[*]') \
    .appName('test') \
    .getOrCreate()
```

```python
spark
```

```python
from pathlib import Path
BASE_DIR = Path.cwd().resolve().parent
RAW_DIR = BASE_DIR / "data" / "raw"
SEASONS = [2023, 2024, 2025, 2026]
```

```python
import re

season_dir = RAW_DIR / str(2023)

pattern = re.compile(r"^matches_\d+.*\.parquet$")
match_files = [str(f) for f in season_dir.iterdir() if pattern.match(f.name)]

if match_files:
    df_2023_m = spark.read.parquet(*match_files)
else:
    print("Không tìm thấy file nào khớp!")
```

```python
df_2023_m.show(1)
```

```python
df_2023_m.createOrReplaceTempView("matches")
```

```python
df_result = spark.sql("""
    SELECT COUNT(*)
    FROM matches
    WHERE id = 435943
""");
```

```python
df_result.show()
```

```python
df_deduplicated = spark.sql("""
    WITH dedup_matches AS (
        SELECT
            *,
            ROW_NUMBER() OVER (
                PARTITION BY id
                ORDER BY last_updated DESC
            ) as row_num
        FROM matches
    )
    SELECT * FROM dedup_matches WHERE row_num = 1
""");
```

```python
df_deduplicated.show(1)
```
